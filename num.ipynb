```python
import pandas as pd
data = {
    'Name': ['mayur', 'swayam', 'sam', 'David', 'Eva'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'HR'],
    'Salary': [70000, 80000, 92000, 52000, 75000],
    'Experience_Years': [3, 5, 7, 2, 5]
}
df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)
print("\n" + "="*30 + "\n")
high_earners = df[df['Salary'] > 70000]
print("--- High Earners (> 60k) ---")
print(high_earners)
print("\n" + "="*30 + "\n")
df['Bonus'] = df['Salary'] * 0.10
print("--- DataFrame with Bonus Column Added ---")
print(df)
print("\n" + "="*30 + "\n")
avg_salary_dept = df.groupby('Department')['Salary'].mean()
print("--- Average Salary by Department ---")
print(avg_salary_dept)
```
